```python
from pathlib import Path
import sys
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))
from src.dataset import get_splits

class_names, splits = get_splits()
summary = pd.DataFrame({
    split: pd.Series([label for _, label in items]).value_counts()
    for split, items in splits.items()
}).fillna(0).astype(int)
summary.index = [class_names[index] for index in summary.index]
summary
```

```python
summary.plot(kind='bar', figsize=(14, 5), title='Class distribution')
plt.ylabel('Images')
plt.xticks(rotation=70)
plt.tight_layout()
plt.show()
```

```python
from PIL import Image

image_paths = sorted((ROOT / 'data').glob('*/*'))
sizes = []
for image_path in image_paths:
    try:
        with Image.open(image_path) as image:
            sizes.append(image.size)
    except (OSError, ValueError):
        pass
pd.Series(sizes, name='size').value_counts().head(10)
```
