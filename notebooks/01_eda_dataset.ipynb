# Dataset EDA

Khảo sát class mapping và split manifest được dùng chung bởi cả ba mô hình.

```python
from pathlib import Path
import sys
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))
from src.dataset import get_splits

class_names, splits = get_splits()
summary = pd.DataFrame({name: pd.Series([label for _, label in items]).value_counts() for name, items in splits.items()}).fillna(0).astype(int)
summary.index = [class_names[index] for index in summary.index]
summary
```

```python
summary.plot(kind='bar', figsize=(14, 5), title='Class distribution by split')
plt.ylabel('Images')
plt.xticks(rotation=70)
plt.tight_layout()
plt.show()
```
